```python
import numpy as np

arr = np.array([1, 2, 3, 6, 4, 5])

# Reverse the array
reverse_arr = arr[::-1]

print("Original array:", arr)
print("Reversed array:", reverse_arr)

```

```python
import numpy as np

x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])

values, counts = np.unique(x, return_counts=True)
max_count = np.max(counts)
most_frequent = values[counts == max_count]

print("Most frequent value:", most_frequent)
print("Frequency:", max_count)

for value in most_frequent:
    print("Indices of", value, ":", np.where(x == value)[0])
```

```python
import numpy as np

ashishbhagat = np.linspace(10, 100, 25)

print("Array:", ashishbhagat)
print("Dimensions:", ashishbhagat.ndim)
print("Shape:", ashishbhagat.shape)
print("Total elements:", ashishbhagat.size)
print("Data type:", ashishbhagat.dtype)
print("Total bytes:", ashishbhagat.nbytes)

transpose = ashishbhagat.reshape(25, 1)
print("Transpose using reshape():")
print(transpose)

print("Transpose using T:")
print(ashishbhagat.T)
```

```python
import numpy as np

ucs420_ashishbhagat = np.array([[10, 20, 30, 40],
                              [50, 60, 70, 80],
                              [90, 15, 20, 35]])

print("Mean:", np.mean(ucs420_ashishbhagat))
print("Median:", np.median(ucs420_ashishbhagat))
print("Max:", np.max(ucs420_ashishbhagat))
print("Min:", np.min(ucs420_ashishbhagat))
print("Unique elements:", np.unique(ucs420_ashishbhagat))

reshaped_ucs420_ashishbhagat = ucs420_ashishbhagat.reshape(4, 3)
print("Reshaped array:")
print(reshaped_ucs420_ashishbhagat)

resized_ucs420_ashishbhagat = np.resize(ucs420_ashishbhagat, (2, 3))
print("Resized array:")
print(resized_ucs420_ashishbhagat)
```
